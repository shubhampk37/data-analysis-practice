```python
import requests

url = "https://cricsheet.org/downloads/ipl_csv2.zip"

response = requests.get(url)
response.raise_for_status()

with open("ipl_csv2.zip", "wb") as f:
  f.write(response.content)

print("Download complete")
```

```python
import zipfile

with zipfile.ZipFile("ipl_csv2.zip", "r") as zip_ref:
    zip_ref.extractall("ipl_data")

print("Extracted successfully.")
```

```python
ball_files = [
    file
    for file in os.listdir("ipl_data")
    if file.endswith(".csv") and not file.endswith("_info.csv")
]

print(f"Number of ball-by-ball files: {len(ball_files)}")
```

```python
matches = []

for file in ball_files:
    df = pd.read_csv(f"ipl_data/{file}")
    matches.append(df)

ball_by_ball = pd.concat(matches, ignore_index=True)

print(f"Combined dataset shape: {ball_by_ball.shape}")
```
