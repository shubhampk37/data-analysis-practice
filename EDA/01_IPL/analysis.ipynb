```python
import requests

url = "https://cricsheet.org/downloads/ipl_csv2.zip"

response = requests.get(url)
response.raise_for_status()

with open("ipl_csv2.zip", "wb") as f:
  f.write(response.content)

print("Download complete")
```

```python
import zipfile

with zipfile.ZipFile("ipl_csv2.zip", "r") as zip_ref:
    zip_ref.extractall("ipl_data")

print("Extracted successfully.")
```

```python
from pathlib import Path

data_path = Path("ipl_data")

files = list(data_path.iterdir())

print("Number of files:", len(files))
print(files[:10])
```

```python
info_files = list(data_path.glob("*_info.csv"))

print("Info files:", len(info_files))
print("\nFirst info file:", info_files[0])

with open(info_files[0], "r", encoding="utf-8") as f:
    for i in range(20):
        print(f.readline().rstrip())
```

```python
import csv

with open(info_files[0], "r", encoding="utf-8") as f:
    rows = list(csv.reader(f))

for row in rows:
    print(row)
```
