```python
import requests

url = "https://cricsheet.org/downloads/ipl_csv2.zip"

response = requests.get(url)
response.raise_for_status()

with open("ipl_csv2.zip", "wb") as f:
  f.write(response.content)

print("Download complete")
```

```python
import zipfile

with zipfile.ZipFile("ipl_csv2.zip", "r") as zip_ref:
    zip_ref.extractall("ipl_data")

print("Extracted successfully.")
```

```python
from pathlib import Path

data_path = Path("ipl_data")

files = list(data_path.iterdir())

print("Number of files:", len(files))
print(files[:10])
```

```python
info_files = list(data_path.glob("*_info.csv"))

print("Info files:", len(info_files))
print("\nFirst info file:", info_files[0])

with open(info_files[0], "r", encoding="utf-8") as f:
    for i in range(20):
        print(f.readline().rstrip())
```

```python
import csv

with open(info_files[0], "r", encoding="utf-8") as f:
    rows = list(csv.reader(f))

for row in rows:
    print(row)
```

```python
import pandas as pd

rows[:10]
```

```python
info_data = {}

for row in rows:
    if row[0] == "info":
        key = row[1]
        value = row[2:]

        if key not in info_data:
            info_data[key] = []

        info_data[key].append(value)

info_data
```

```python
matches = []

for info_file in info_files:
    with open(info_file, "r", encoding="utf-8") as f:
        rows = list(csv.reader(f))

    info_data = {}

    for row in rows:
        if len(row) >= 3 and row[0] == "info":
            key = row[1]
            value = row[2:]

            if key not in info_data:
                info_data[key] = []

            info_data[key].append(value)

    match_data = {
        "match_id": info_data["match_id"][0][0],
        "season": info_data["season"][0][0],
        "date": info_data["date"][0][0],
        "team_1": info_data["team"][0][0],
        "team_2": info_data["team"][1][0],
        "venue": info_data["venue"][0][0],
        "city": info_data["city"][0][0],
        "toss_winner": info_data["toss_winner"][0][0],
        "toss_decision": info_data["toss_decision"][0][0],
        "player_of_match": info_data.get("player_of_match", [[None]])[0][0],
        "winner": info_data.get("winner", [[None]])[0][0],
        "winner_runs": info_data.get("winner_runs", [[None]])[0][0],
        "winner_wickets": info_data.get("winner_wickets", [[None]])[0][0],
        "outcome": info_data.get("outcome", [[None]])[0][0]
    }

    matches.append(match_data)

df = pd.DataFrame(matches)

df["date"] = pd.to_datetime(df["date"])
df["winner_runs"] = pd.to_numeric(df["winner_runs"])
df["winner_wickets"] = pd.to_numeric(df["winner_wickets"])

print("Dataset shape:", df.shape)
df.head()
```

```python
team_appearances = (
    df["team_1"].value_counts()
    .add(df["team_2"].value_counts(), fill_value=0)
    .sort_values(ascending=False)
)

team_appearances
```

```python
team_wins = df["winner"].value_counts()

team_wins
```

```python
team_summary = pd.DataFrame({
    "matches_played": team_appearances,
    "matches_won": team_wins
}).fillna(0)

team_summary["matches_won"] = team_summary["matches_won"].astype(int)

team_summary
```

```python
team_summary["win_percentage"] = (
    team_summary["matches_won"] / team_summary["matches_played"] * 100
).round(2)

team_summary.sort_values("win_percentage", ascending=False)
```
